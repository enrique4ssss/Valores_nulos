```python
import pandas as pd
import numpy as np

# 1. Cargar el dataset original
df_raw = pd.read_csv('listings.csv')

# 2. Lista de las 50 columnas seleccionadas
cols_50 = [
    # Identificación / Básico (5)
    'id', 'name', 'description', 'neighborhood_overview', 'instant_bookable',
    
    # Anfitrión (11)
    'host_id', 'host_name', 'host_since', 'host_location', 'host_about', 
    'host_response_time', 'host_response_rate', 'host_acceptance_rate', 
    'host_is_superhost', 'host_listings_count', 'host_identity_verified',
    
    # Ubicación y Propiedad (12)
    'neighbourhood_cleansed', 'neighbourhood_group_cleansed', 'latitude', 'longitude', 
    'property_type', 'room_type', 'accommodates', 'bathrooms_text', 
    'bedrooms', 'beds', 'amenities', 'license',
    
    # Precios, Reglas y Disponibilidad (11)
    'price', 'minimum_nights', 'maximum_nights', 'has_availability', 
    'availability_30', 'availability_60', 'availability_90', 'availability_365', 
    'estimated_occupancy_l365d', 'estimated_revenue_l365d', 'price_quote_price_per_night',
    
    # Reseñas y Métricas (11)
    'number_of_reviews', 'number_of_reviews_ltm', 'reviews_per_month', 
    'first_review', 'last_review', 'review_scores_rating', 'review_scores_accuracy', 
    'review_scores_cleanliness', 'review_scores_checkin', 'review_scores_communication', 
    'review_scores_value'
]

# Filtrar el DataFrame a solo las 50 columnas
df = df_raw[cols_50].copy()

# ---------------------------------------------------------
# 3. Limpieza previa de formatos (Precios y Porcentajes)
# ---------------------------------------------------------
if df['price'].dtype == 'object':
    df['price'] = df['price'].astype(str).str.replace('$', '', regex=False).str.replace(',', '', regex=False)
    df['price'] = pd.to_numeric(df['price'], errors='coerce')

for col in ['host_response_rate', 'host_acceptance_rate']:
    if df[col].dtype == 'object':
        df[col] = df[col].astype(str).str.replace('%', '', regex=False)
        df[col] = pd.to_numeric(df[col], errors='coerce')

# ---------------------------------------------------------
# 4. Tratamiento e Imputación de Nulos
# ---------------------------------------------------------

# A. Categóricas / Texto
cat_constante = ['description', 'neighborhood_overview', 'host_about', 'license', 'neighbourhood_group_cleansed']
for col in cat_constante:
    df[col] = df[col].fillna('Sin información')

cat_moda = ['host_name', 'host_location', 'host_response_time', 'host_is_superhost', 
            'host_identity_verified', 'neighbourhood_cleansed', 'property_type', 
            'room_type', 'bathrooms_text', 'has_availability', 'instant_bookable',
            'host_since', 'first_review', 'last_review']

# Validación para evitar KeyError si una columna está 100% vacía
for col in cat_moda:
    moda = df[col].mode()
    if not moda.empty:
        df[col] = df[col].fillna(moda[0])
    else:
        df[col] = df[col].fillna('Sin información')

# B. Variables DISCRETAS (Conteos enteros) -> Imputación por MEDIANA
cols_discretas = [
    'accommodates', 'bedrooms', 'beds', 'host_listings_count',
    'minimum_nights', 'maximum_nights', 'availability_30', 'availability_60', 
    'availability_90', 'availability_365', 'number_of_reviews', 
    'number_of_reviews_ltm', 'estimated_occupancy_l365d'
]

for col in cols_discretas:
    mediana_val = df[col].median()
    # Si la columna entera fuera nula, se usa 0 como respaldo
    if pd.isna(mediana_val):
        mediana_val = 0
    df[col] = df[col].fillna(mediana_val)

# C. Variables CONTINUAS (Decimales / Precios / Promedios) -> Imputación por MEDIA
cols_continuas = [
    'price', 'price_quote_price_per_night', 'estimated_revenue_l365d', 
    'latitude', 'longitude', 'host_response_rate', 'host_acceptance_rate', 
    'reviews_per_month', 'review_scores_rating', 'review_scores_accuracy', 
    'review_scores_cleanliness', 'review_scores_checkin', 
    'review_scores_communication', 'review_scores_value'
]

for col in cols_continuas:
    media_val = df[col].mean()
    # Si la columna entera fuera nula, se usa 0.0 como respaldo
    if pd.isna(media_val):
        media_val = 0.0
    df[col] = df[col].fillna(media_val)

# ---------------------------------------------------------
# 5. Verificación y Exportación
# ---------------------------------------------------------
print("Total de nulos restantes en el dataset:", df.isnull().sum().sum())

# Exportar archivo final sin nulos
df.to_csv('Listings_50_Limpio.csv', index=False)
```

```python
import pandas as pd
import numpy as np

# 1. Cargar el dataset original
df_raw = pd.read_csv('listings.csv')

# 2. Lista de las 50 columnas seleccionadas
cols_50 = [
    # Identificación / Básico (5)
    'id', 'name', 'description', 'neighborhood_overview', 'instant_bookable',
    
    # Anfitrión (11)
    'host_id', 'host_name', 'host_since', 'host_location', 'host_about', 
    'host_response_time', 'host_response_rate', 'host_acceptance_rate', 
    'host_is_superhost', 'host_listings_count', 'host_identity_verified',
    
    # Ubicación y Propiedad (12)
    'neighbourhood_cleansed', 'neighbourhood_group_cleansed', 'latitude', 'longitude', 
    'property_type', 'room_type', 'accommodates', 'bathrooms_text', 
    'bedrooms', 'beds', 'amenities', 'license',
    
    # Precios, Reglas y Disponibilidad (11)
    'price', 'minimum_nights', 'maximum_nights', 'has_availability', 
    'availability_30', 'availability_60', 'availability_90', 'availability_365', 
    'estimated_occupancy_l365d', 'estimated_revenue_l365d', 'price_quote_price_per_night',
    
    # Reseñas y Métricas (11)
    'number_of_reviews', 'number_of_reviews_ltm', 'reviews_per_month', 
    'first_review', 'last_review', 'review_scores_rating', 'review_scores_accuracy', 
    'review_scores_cleanliness', 'review_scores_checkin', 'review_scores_communication', 
    'review_scores_value'
]

# Filtrar el DataFrame a solo las 50 columnas
df = df_raw[cols_50].copy()

# ---------------------------------------------------------
# 3. LIMPIEZA DE FORMATOS Y CONVERSIÓN A NUMÉRICO
# ---------------------------------------------------------

# Lista de todas las columnas que deben ser numéricas (discretas y continuas)
cols_discretas = [
    'accommodates', 'bedrooms', 'beds', 'host_listings_count',
    'minimum_nights', 'maximum_nights', 'availability_30', 'availability_60', 
    'availability_90', 'availability_365', 'number_of_reviews', 
    'number_of_reviews_ltm', 'estimated_occupancy_l365d'
]

cols_continuas = [
    'price', 'price_quote_price_per_night', 'estimated_revenue_l365d', 
    'latitude', 'longitude', 'host_response_rate', 'host_acceptance_rate', 
    'reviews_per_month', 'review_scores_rating', 'review_scores_accuracy', 
    'review_scores_cleanliness', 'review_scores_checkin', 
    'review_scores_communication', 'review_scores_value'
]

# Limpiar caracteres especiales ($ % ,) y fuerza la conversión a float/int
for col in cols_discretas + cols_continuas:
    if col in df.columns:
        df[col] = df[col].astype(str).str.replace('$', '', regex=False) \
                                    .str.replace('%', '', regex=False) \
                                    .str.replace(',', '', regex=False) \
                                    .str.strip()
        # errors='coerce' convierte cualquier texto no válido a NaN para poder calcular medias/medianas
        df[col] = pd.to_numeric(df[col], errors='coerce')

# ---------------------------------------------------------
# 4. TRATAMIENTO E IMPUTACIÓN DE NULOS
# ---------------------------------------------------------

# A. Categóricas / Texto
cat_constante = ['description', 'neighborhood_overview', 'host_about', 'license', 'neighbourhood_group_cleansed']
for col in cat_constante:
    df[col] = df[col].fillna('Sin información')

cat_moda = ['host_name', 'host_location', 'host_response_time', 'host_is_superhost', 
            'host_identity_verified', 'neighbourhood_cleansed', 'property_type', 
            'room_type', 'bathrooms_text', 'has_availability', 'instant_bookable',
            'host_since', 'first_review', 'last_review']

for col in cat_moda:
    moda = df[col].mode()
    if not moda.empty:
        df[col] = df[col].fillna(moda[0])
    else:
        df[col] = df[col].fillna('Sin información')

# B. Variables DISCRETAS -> Imputación por MEDIANA
for col in cols_discretas:
    mediana_val = df[col].median()
    if pd.isna(mediana_val):
        mediana_val = 0
    df[col] = df[col].fillna(mediana_val)

# C. Variables CONTINUAS -> Imputación por MEDIA
for col in cols_continuas:
    media_val = df[col].mean()
    if pd.isna(media_val):
        media_val = 0.0
    df[col] = df[col].fillna(media_val)

# ---------------------------------------------------------
# 5. VERIFICACIÓN Y EXPORTACIÓN
# ---------------------------------------------------------
print("Total de nulos restantes en el dataset:", df.isnull().sum().sum())

# Exportar archivo final sin nulos
df.to_csv('Listings_50_Limpio.csv', index=False)
```
